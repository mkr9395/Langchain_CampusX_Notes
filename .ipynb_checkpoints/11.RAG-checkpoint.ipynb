### A pretrined model is trained on a smaller, domain specific dataset

### 🔹 What is Fine-Tuning of LLMs?

* Imagine an **LLM (Large Language Model)** like **GPT, LLaMA, Falcon** as a **general brain** trained on *huge amounts of internet data*.


* It knows a *lot about everything*, but it’s **not specialized**.




**Fine-tuning** = Taking that **general brain** and giving it **specialized training** on your own data so it becomes an **expert** in your domain/task.



Think of a doctor:

* **General training (LLM pretraining):** learns all medicine basics.
  
  
  
* **Specialization (fine-tuning):** trains only on *heart diseases* → becomes a **cardiologist**.



Similarly:

* **LLM before fine-tuning:** knows general English/text.


* **LLM after fine-tuning:** knows *legal language*, *medical reports*, *banking transactions*, etc.

### 🔹 Why Do We Fine-Tune?

* To **customize** the LLM for a **specific domain** (finance, healthcare, law).


* To **improve performance** on a **specific task** (sentiment analysis, summarization, classification).


* To make it **more aligned** with your company’s style, rules, or data.

### 🔹 Example

* Base LLM: GPT-like model.


* Fine-tune it with **bank transaction data** → Now it can **categorize transactions** into groceries, rent, utilities, etc.

---

👉 In short:
**Fine-tuning = teaching a big general model to become a focused expert on *your* data/task.**

### 🔹 Types of Fine-Tuning

1. **Full fine-tuning** – retrain *all model parameters* (very costly).


2. **Parameter-efficient fine-tuning (PEFT)** – only train *a few extra layers* (cheap & popular).

   * Example: **LoRA (Low-Rank Adaptation)** → adds small adapters to the LLM.


### 🔹 1. **Full Fine-Tuning**

* You update **all parameters** of the model on your dataset.
* Example: Training GPT-2 on movie reviews for sentiment analysis.
* **Pros:** Best performance if enough data.
* **Cons:** Very expensive (needs lots of GPU, time, and data).

---

### 🔹 2. **Parameter-Efficient Fine-Tuning (PEFT)**

Instead of updating all parameters, you update only a **small subset**. Popular for LLMs because they’re huge.

### a) **LoRA (Low-Rank Adaptation)**

* Adds small trainable weight matrices to existing layers, freezing the original model.
* Efficient, widely used in practice.
* Example: Fine-tuning LLaMA-2 for customer support with just a few GB of data.

### b) **Adapters**

* Small “adapter” layers are added between transformer layers, and only those are trained.
* Keeps original model intact.
* Example: Training multilingual adapters for translation.

### c) **Prefix / Prompt Tuning**

* You **learn special tokens (prompts)** that guide the model instead of changing weights.
* Model stays mostly frozen.
* Example: Learn a soft prompt to make GPT generate medical reports.

### d) **BitFit**

* Only train the **bias terms** in each layer.
* Very lightweight.

---

### 🔹 3. **Instruction Tuning**

* Train the model on a dataset of **instruction → response pairs**.
* Helps align the LLM with following user instructions (like ChatGPT).

---

### 🔹 4. **Domain-Specific Fine-Tuning**

* Train on a domain dataset (finance, law, healthcare, etc.).
* Example: Fine-tuning GPT on legal contracts for legal Q\&A.

---

### 🔹 5. **RLHF (Reinforcement Learning from Human Feedback)**

* After supervised fine-tuning, the model is further trained using **human preference signals**.
* Used in **ChatGPT** to make outputs more helpful, safe, and aligned.

✅ **In simple terms:**

* **Full fine-tuning** = retrain the whole brain.


* **PEFT methods (LoRA, Adapters, Prefix tuning, BitFit)** = retrain small parts or add small “plugins” to the brain.


* **Instruction tuning** = train the brain to follow tasks better.


* **RLHF** = adjust based on human likes/dislikes.

## RAG vs FineTuning

### **1. Fine-tuning**

* You **train the model itself** on your domain data.


* The model **absorbs knowledge into its weights**.


* Once trained, it **remembers** the domain info without needing an external database.

✅ Best when:

* You need the model to **learn a new style, tone, or reasoning pattern**.


* You have a **special dataset** (e.g., legal contracts, financial reports, medical notes).


* The task is **repetitive & structured** (classification, summarization, chatbot with fixed tone).

⚠️ Limitation:

* Expensive (training cost + compute).


* Hard to update (if new data comes, you must fine-tune again).


* Can overfit if data is small.

### **2. RAG (Retrieval-Augmented Generation)**

* You **don’t change the model’s weights**.


* Instead, you **connect the model to an external knowledge base** (like a database, vector store, or documents).


* At query time, the system **retrieves relevant info** and passes it to the model to generate a response.



✅ Best when:

* Knowledge changes **frequently** (e.g., latest financial filings, company policies, news).


* You have **lots of documents** but don’t want to retrain.


* You want **cost-effective updates** (just update your vector database, no training).



⚠️ Limitation:


* Model doesn’t truly “learn” the knowledge → it just **looks it up**.


* Performance depends on the **retriever quality**.


* Not great if you want the model to adopt a **new writing style or reasoning ability**.

### 🔹 Analogy

* **Fine-tuning** = Sending the LLM to *university again* to **learn new knowledge permanently**.


* **RAG** = Giving the LLM a **Google search engine** — it doesn’t memorize everything, but can **look things up on the fly**.

### 🔹 Example

* **Fine-tuning:** Train GPT on **bank transactions** → it *automatically knows how to classify transactions* without external lookup.


* **RAG:** Store **bank policies and descriptions** in a database → GPT retrieves the right policy whenever a user asks a question.



👉 **Rule of Thumb**:

* Use **RAG** if your data **changes often** or is **too big** to train on.



* Use **fine-tuning** if you need the model to **permanently learn a skill, tone, or reasoning style**.

## RAG

- In RAG we send the context document and the query both to the LLM during prompting.


- we don't send the entire document, only the context.


- RAG (Retrieval-Augmented Generation) is a method where a Large Language Model (LLM) is connected to an external knowledge source (like a database, vector store, or document collection). Instead of relying only on what it was trained on, the LLM can retrieve relevant information at query time and then use it to generate more accurate and up-to-date responses.

#### PROMPT + PARAMETRIC KNOWLEDGE = RAG

### Exampel : 

### **How it works (in simple terms)**

Think of RAG as giving the LLM a **"smart notebook"** it can open every time you ask a question:

1. **User asks a question** → Example: *"What are the key points in Tesla’s latest 10-K filing?"*


2. **Retriever kicks in** → It searches the connected knowledge base (e.g., PDFs, vector DB, documents) for the most relevant chunks of text.


3. **LLM + retrieved info** → The retrieved documents are passed into the LLM’s context window along with the question.


4. **Final Answer** → The LLM combines its reasoning ability + the retrieved context to generate a fact-based, updated answer.

## **Simple analogy**

* **Without RAG** → The LLM is like a student answering only from memory (which may be outdated or incomplete).


* **With RAG** → The LLM is like a student who checks their textbooks/notes before answering.

## Understanding RAG 

### **core steps of RAG :**

### **1. Indexing** (Preparing the knowledge base)

* This is the step where you **organize external data** (documents, PDFs, websites, knowledge graphs, etc.) into a format the model can search.
* Typically done by:

  * Splitting documents into chunks (so the model doesn’t get overloaded with huge text).
  * Converting chunks into embeddings (vector representations of meaning).
  * Storing them in a **vector database** (like Pinecone, Weaviate, FAISS, Milvus).

👉 **Analogy**: Like creating an index at the back of a book so you can find topics quickly.

### 4 Steps in Indexing : 

##### External Knowledge base in set

---

### **2. Retrieval** (Finding relevant knowledge)

* When a user asks a question, RAG searches the vector database to **retrieve the most relevant document chunks**.
* This uses **semantic similarity** (meaning-based search) rather than keyword search.

👉 **Analogy**: Instead of flipping through every page, you use the book’s index to instantly find the right sections.

##### Most relevant chunks generated

---
### **3. Augmentation** (Feeding retrieved info to the LLM)

* The retrieved chunks are **added to the model’s input prompt** along with the user’s query.
* This step “augments” the LLM with **fresh, domain-specific, or proprietary knowledge** that wasn’t part of its original training.

👉 **Analogy**: The student takes notes from the textbook (retrieved chunks) and attaches them to the exam question before answering.

##### Prompt is generated to be asked to the LLM

---

### **4. Generation** (Producing the final answer)

* The LLM now has both:

  * The **user’s question**, and
  * The **retrieved context**.
* It combines reasoning + retrieved knowledge to generate a **context-aware, factual, and useful response**.

👉 **Analogy**: The student reads the question, checks notes from the textbook, and writes a clear final answer.

---

### Putting it all together (Flow of RAG):

1. **Indexing** → Store knowledge in a searchable way.


2. **Retrieval** → Find the most relevant chunks when a query comes.


3. **Augmentation** → Attach those chunks to the LLM’s prompt.


4. **Generation** → LLM produces a better, fact-based answer.
