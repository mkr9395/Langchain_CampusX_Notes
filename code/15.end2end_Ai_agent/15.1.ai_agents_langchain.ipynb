```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
import requests
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
```

```python

# Step 1.1. creating the 1st tool -> search tool
search_tool = DuckDuckGoSearchRun()
```

```python
# Step 1.2. 2nd tool to fetch weather data
@tool
def get_weather_data(city : str) -> str:
    """
    This function fetches the current weather data for a given city
    """
    url = f'https://api.weatherstack.com/current?access_key=185af98cdd4e516c3774aa94f15c2c50&query={city}'
    
    response = requests.get(url)
    
    return response.json()
```

```python
llm = ChatOpenAI()
```

```python
from langchain.agents import create_react_agent, AgentExecutor
from langchain import hub
```

```python
# Step 2: Pull the ReAct prompt from LangChain Hub
prompt = hub.pull("hwchase17/react")  # pulls the standard ReAct agent prompt
```

```python
# Step 3: Create the ReAct agent manually with the pulled prompt

agent = create_react_agent(
    llm = llm, # which llm it needs for reasoning capabilities
    tools = [search_tool, get_weather_data], # which tool acces the agent will get
    prompt = prompt # the prompt needed
)
```

```python
# Step 4: Wrap it with AgentExecutor

agent_executor = AgentExecutor(
    agent = agent, # agent will be provided
    tools = [search_tool, get_weather_data], # specify the tools again
    verbose=True
)
```

```python
# Step 5: Invoke
response = agent_executor.invoke({"input":"Find the capital of Madhya Pradesh, the find it's current weather condition"})

print(response)
```
